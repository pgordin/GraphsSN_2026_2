# Graph functions


```python
def print_matrix(vertices, matrix):
  """
  Print a graph given by  the matrix.
  """
  n = len(matrix)   # the order of thew graph
  # take the proper names of vertices
  if (vertices is not None) and (len(vertices) == n):
    vv = vertices
  else:
    vv = range(1, n+1)  ## numbers {1..n}
  for i in range(n):
    print(vv[i], ': ', end='', sep = '')
    for j in range(n):
      if matrix[i][j]:
        print(vv[j], end=' ')
    print()


def print_graph(graph):
  """
  Prints a graph given by an adjacency list (a dictionary).
  """
  for v in graph:
    print(v, ': ', end='', sep = '')
    for u in graph[v]:
      print(u, end=' ')
    print()
```

# Modyfying a graph

```python
def add_vertex(graph, vertex):
  """
  Adds a vertex to a graph.
  """
  if vertex not in graph:
    graph[vertex] = []
```

```python
def add_arc(graph, arc):
  """
  Adds an arc to a simple (di)graph.
  """
  u, v = arc
  add_vertex(graph, u)
  add_vertex(graph, v)
  if v not in graph[u]:
    graph[u].append(v)
```

```python
def add_edge(graph, edge):
  """
  Adds an edge to a simple graph. Loops are not allowed.
  """
  u, v = edge
  if (u == v):
    raise ValueError("Loops are not allowed.")
  add_vertex(graph, u)
  add_vertex(graph, v)
  if v not in graph[u]:
    graph[u].append(v)
  if u not in graph[v]:
    graph[v].append(u)
```

# Random graph generator with $G(n,p)$ model.

```python
from random import random, seed
# for repeatness
seed(2026)
```

```python
# prepare graph G(10, 1/3)
n = 10
p = 1/3
random_graph = {}
for i in range(1, n+1):
  add_vertex(random_graph, i)
  for j in range(1, i):
    if random() < p:
        add_edge(random_graph, (i, j))
```

```python
print_graph(random_graph)
```
