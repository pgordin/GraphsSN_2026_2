**Graph Functions**

```python
import random
import networkx as nx
import numpy as np

random.seed(42)
```

```python
def print_matrix(vertices, matrix):
  """
  Print a graph given by  the matrix.
  """
  n = len(matrix) 
  if (vertices is not None) and (len(vertices) == n):
    vv = vertices
  else:
    vv = range(1, n+1)  ## numbers {1..n}
  for i in range(n):
    print(vv[i], ': ', end='', sep = '')
    for j in range(n):
      if matrix[i][j]:
        print(vv[j], end=' ')
    print()

  
```

```python
def print_graph(graph):
  """
  Prints a graph given by an adjacency list (a dictionary).
  """
  for v in graph:
    print(v, ': ', end='', sep = '')
    for u in graph[v]:
      print(u, end=' ')
    print()
```

```python
def add_vertex(graph, vertex):
  """
  adds a vertex to a graph
  """
  if vertex not in graph:
    graph[vertex] = []
```

```python
def add_edge(graph, edge):
  """
  adds an edge to a graph (simple, undirected graph) (loops are not allowed)
  """
  u, v = edge
  add_vertex(graph, u)
  add_vertex(graph, v)
  if u == v:
    raise ValueError("loops are not allowed")
  if v not in graph[u]:
    graph[u].append(v)
  if u not in graph[v]:
    graph[v].append(u)
```

```python
def random_graphn(n, p):
    graph = {}
    for i in range(1, n+1):
        add_vertex(graph, i)
        for j in range(1, i):
            if random.random() <= p:
                add_edge(graph, (i, j))
    return graph
```

```python
def graph_to_matrix(graph):
    '''
    1. Take a list of vertices names
    2. Index them (e.g.: A: 0, B: 1)
    3. Create zero-matrix
    4. Using indexes from step 2 set values of connected vertices to 1
    '''
    n = len(graph)
    vertices = list(graph.keys())
    index = {v: i for i, v in enumerate(vertices)}
    matrix = [[0 for _ in range(n)] for _ in range(n)]
    for v in vertices:
        for neighbor in graph[v]:
            matrix[index[v]][index[neighbor]] = 1
    return vertices, matrix
```

```python
def matrix_to_graph(vertices, matrix):
    graph = {v: [] for v in vertices}
    n = len(vertices)
    for i in range(n):
        for j in range(n):
            if matrix[i][j] == 1:
                graph[vertices[i]].append(vertices[j])
    return graph
```

```python
def cycle(n):
    graph = {}
    for i in range(n):
        add_edge(graph, (i+1, (i+1) % n + 1))
    return graph
```
