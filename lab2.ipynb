**Graph functions**

```python
def print_matrix(vertices, matrix):
  """
  Print a graph given by  the matrix.
  """
  n = len(matrix)   # the order of thew graph
  # take the proper names of vertices
  if (vertices is not None) and (len(vertices) == n):
    vv = vertices
  else:
    vv = range(1, n+1)  ## numbers {1..n}
  for i in range(n):
    print(vv[i], ': ', end='', sep = '')
    for j in range(n):
      if matrix[i][j]:
        print(vv[j], end=' ')
    print()
```

```python
def print_graph(graph):
  """
  Prints a graph given by an adjacency list (a dictionary).
  """
  for v in graph:
    print(v, ': ', end='', sep = '')
    for u in graph[v]:
      print(u, end=' ')
    print()
```

**Modyfing the graphs**

```python
def add_vertex(graph, vertex):
  """
  Adds a vertex to a graph.
  """
  if vertex not in graph:
    graph[vertex] = []
```

```python
def add_arc(graph, arc):
  """
  Adds an arc to a simple (di)graph.
  """
  u, v = arc
  add_vertex(graph, u)
  add_vertex(graph, v)
  if v not in graph[u]:
    graph[u].append(v)
```

```python
def add_edge(graph, edge):
  """
  Adds an edge to a simple graph. Loops are not allowed.
  """
  u, v = edge
  if (u == v):
    raise ValueError("Loops are not allowed.")
  add_vertex(graph, u)
  add_vertex(graph, v)
  if v not in graph[u]:
    graph[u].append(v)
  if u not in graph[v]:
    graph[v].append(u)
```

```python
def graph_from_edges(filename, directed = 0):
  """
  Read the graph from a file, that each line contains ether the description of
  an edge (two words - pair of vertices) or the description of a vertex (one word).
  The result is a graph in a dictionary format.
  filename contains the whole path to the file
  """
  graph = {}
  file = open(filename, "r")    # open the file for reading
  for line in file:
    words = line.strip().split()  # spilt the line into words
    if len(words)==1:             # one word - the vertex description
      add_vertex(graph, words[0])
    elif len(words) >= 2:          # two words - the edge description
      if directed == 0:
        add_edge(graph, (words[0], words[1]))
      else:
        add_arc(graph, (words[0], words[1]))
  file.close()
  return graph
```

```python
def graph_to_neighbourlist(graph, filename):
  """
  Write a graph as a file in the neighbourlist format
  """
  file = open(filename, "w")
  for v in graph:
    neigh_list = f"{v}:"
    for u in graph[v]:
      neigh_list += f" {u}"
    neigh_list += "\n"
    file.write(neigh_list)
  file.close()
```

```python
def graph_to_edges(graph, filename):
  """
  Write a graph as an edges-format file, without duplicate edges.
  Isolated vertices are written on their own line.
  """
  written = set()
  with open(filename, "w") as file:
    for v in graph:
      if not graph[v]:
        file.write(f"{v}\n")
        continue

      for u in graph[v]:
        edge = frozenset((v, u))
        if edge not in written:
          file.write(f"{v} {u}\n")
          written.add(edge)
```

```python
def graph_to_edges_duplicates(graph, filename):
  """
  Write a graph as a file in the edges format
  """
  file = open(filename, "w")
  for v in graph:
    for u in graph[v]:
      file.write(f"{v} {u}\n")
  file.close()
```

```python
def graph_from_neighbourlist(filename):
  """
  Read the graph from a file, that each line contains vertex and all edges
  from this vertes
  The result is a graph in a dictionary format.
  filename contains the whole path to the file
  """
  graph = {}
  file = open(filename, "r")    # open the file for reading
  for line in file:
    vertex, conn = line.strip().split(":")
    if conn == "":
      add_vertex(graph, vertex)
    else:
      neighbours = conn.strip().split()
      add_vertex(graph, vertex)
      for neighbour in neighbours:
        add_edge(graph, (vertex, neighbour))
  file.close()
  return graph
```
